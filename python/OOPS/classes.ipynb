```python
class Student:
    school = "AMU"                      # class attribute (shared by all)

    def __init__(self, name, age):      # constructor
        self.name = name                # instance attributes
        self.age = age

    def introduce(self):                # instance method
        return f"I'm {self.name}, {self.age} years old."

s1 = Student("Asha", 20)
print(s1.introduce())    # I'm Asha, 20 years old.
print(s1.school)         # AMU
```

```python
class animal:
    def __init__(self,name):
        self.name=name
    def eat(self):
        print(f"{self.name} is eeting")
    def sleep(self):
        print(f"{self.name}is sleeping")

class dog(animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed
    def bark(self):
        print(f"{self.name} is barking")
        
d=dog("Fido","Labrador")
d.bark()
d.eat()
d.sleep()
print(d.breed)
        
```

```python
class shape:
    def area(self):
        return 0
class rectangle(shape):
    def __init__(self,w,h):
        self.w=w
        self.h=h
        
    def area(self):
        return self.w * self.h
    
class circle(shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14 * self.r**2
shapes=[rectangle(2,3),circle(4)]
for s in shapes:
    print(s.area())
```
