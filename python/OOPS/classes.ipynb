```python
class Student:
    school = "AMU"                      # class attribute (shared by all)

    def __init__(self, name, age):      # constructor
        self.name = name                # instance attributes
        self.age = age

    def introduce(self):                # instance method
        return f"I'm {self.name}, {self.age} years old."

s1 = Student("Asha", 20)
print(s1.introduce())    # I'm Asha, 20 years old.
print(s1.school)         # AMU
```
