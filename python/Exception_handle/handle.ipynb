```python
#raise concept 
class bank:
    def __init__(self,balance):
        self.balance=balance
    
    def withdraw(self,amount):
        if amount<0:
            raise Exception("amount cannot be negative")
        if self.balance < amount:
            raise Exception("insufficient balance")
        self.balance=self.balance-amount
obj=bank(10000)
try:
    obj.withdraw(5000)
except Exception as e:
    print(e)
else:
    print(obj.balance)

```

```python
#creating custom exception
class MyException(Exception):
    def __init__(self,msg):
        print(msg)
        
class bank:
    def __init__(self,balance):
        self.balance=balance
    
    def withdraw(self,amount):
        if amount<0:
            raise MyException("amount cannot be negative")
        if self.balance < amount:
            raise MyException("insufficient balance")
        self.balance=self.balance-amount
obj=bank(10000)
try:
    obj.withdraw(5000)
except MyException as e:
    pass
else:
    print(obj.balance)


```

```python
#why to use the custom exception and where to use it?
#example 
class SecurityError(Exception):
    def __init__(self,msg):
        print(msg)
    def logout(self):
        print("logout")
        
class google:
    def __init__(self,name,emai,password,device):
        self.name=name
        self.email=emai
        self.password=password
        self.device=device
        
    def login(self,email,password,device):
        if device!=self.device:
            raise SecurityError("alert:device not matched")
        if email==self.email and password==self.password:
            print("login success")
        else:
            print("login failed")
            

obj=google("ram","ram@gmail.com","123456","phone")
try:
    obj.login("ram@gmail.com","123456","phone")
except SecurityError as e:
    e.logout()
else:
    print(obj.name)
finally:
    print("connection closed")
    
        
```
